# 5. Data Cleaning

This section cleans the dataset by handling missing values, removing duplicate records, converting data types, and verifying the final cleaned dataset.

### 5.1 Handle Missing Values

Missing values are handled using appropriate placeholder values so that the dataset can be used for further analysis without losing important records.

```python
df_clean = df.copy()

df_clean['director'] = df_clean['director'].fillna('Not Available')
df_clean['cast'] = df_clean['cast'].fillna('Not Available')
df_clean['country'] = df_clean['country'].fillna('Not Available')
df_clean['date_added'] = df_clean['date_added'].fillna('Not Available')
df_clean['rating'] = df_clean['rating'].fillna('Not Rated')
df_clean['duration'] = df_clean['duration'].fillna('Not Available')

print("Missing values after cleaning:")
print(df_clean.isnull().sum())
```

### 5.2 Remove Duplicate Records

Duplicate records are removed from the cleaned dataset to ensure that each record is counted only once during analysis.

```python
df_clean = df_clean.drop_duplicates()

print("Number of Duplicate Records:", df_clean.duplicated().sum())
print("Dataset Shape:", df_clean.shape)
```

### 5.3 Data Type Conversion

The `date_added` column is converted to a datetime format to support time-based analysis.

```python
df_clean['date_added'] = pd.to_datetime(
    df_clean['date_added'],
    errors='coerce'
)

df_clean['date_added'] = df_clean['date_added'].fillna(
    pd.Timestamp('1900-01-01')
)

print(df_clean.dtypes)
```

### 5.4 Verify Cleaned Dataset

The cleaned dataset is verified by checking its shape, remaining missing values, and duplicate records.

```python
print("Final Dataset Shape:", df_clean.shape)
print("Total Missing Values:", df_clean.isnull().sum().sum())
print("Total Duplicate Records:", df_clean.duplicated().sum())
```

# 7. Sentiment Analysis

This section analyzes the sentiment of Netflix content descriptions and classifies them as Positive, Neutral, or Negative based on their sentiment polarity.

### 7.1 Prepare Description Data

The description column is prepared for sentiment analysis by ensuring that all descriptions contain usable text.

```python
from textblob import TextBlob

df_sentiment = df_clean[['release_year', 'description']].copy()

df_sentiment['description'] = df_sentiment['description'].fillna('')

print("Total Records:", len(df_sentiment))
display(df_sentiment.head())
```

### 7.2 Calculate Sentiment

Sentiment polarity is calculated for each content description using TextBlob. The polarity value ranges from negative to positive sentiment.

```python
df_sentiment['polarity'] = df_sentiment['description'].apply(
    lambda text: TextBlob(text).sentiment.polarity
)

print(df_sentiment[['release_year', 'polarity']].head())
```

### 7.3 Classify Positive, Neutral and Negative

The sentiment polarity values are classified into three categories: Positive, Neutral, and Negative.

```python
def classify_sentiment(polarity):
    if polarity > 0:
        return 'Positive'
    elif polarity < 0:
        return 'Negative'
    else:
        return 'Neutral'

df_sentiment['sentiment'] = df_sentiment['polarity'].apply(
    classify_sentiment
)

print(df_sentiment['sentiment'].value_counts())
```

### 7.4 Sentiment by Release Year

**Question:** How does the sentiment of Netflix content descriptions vary across release years?

The number of Positive, Neutral, and Negative descriptions is analyzed across release years.

```python
sentiment_year = (
    df_sentiment[
        df_sentiment['release_year'] > 2005
    ]
    .groupby(['release_year', 'sentiment'])
    .size()
    .reset_index(name='Total Count')
)

sentiment_pivot = sentiment_year.pivot(
    index='release_year',
    columns='sentiment',
    values='Total Count'
).fillna(0)

print(sentiment_pivot.head())
```

### 7.5 Sentiment Visualization

The distribution of Positive, Neutral, and Negative sentiments across release years is visualized using an interactive Plotly chart.

```python
sentiment_long = sentiment_pivot.reset_index().melt(
    id_vars='release_year',
    var_name='Sentiment',
    value_name='Total Count'
)

fig = px.bar(
    sentiment_long,
    x='release_year',
    y='Total Count',
    color='Sentiment',
    title='Sentiment Analysis of Netflix Content by Release Year',
    labels={
        'release_year': 'Release Year',
        'Total Count': 'Number of Titles'
    }
)

fig.show()
```

# 8. Interactive Dashboard

### 8.1 Dashboard Overview

An interactive dashboard is created directly inside Google Colab using Python, Plotly, and ipywidgets. The dashboard allows users to filter Netflix content and explore key statistics and visualizations interactively.

```python
import ipywidgets as widgets
from IPython.display import display, clear_output

print("Interactive Dashboard Libraries Loaded Successfully")
```

### 8.2 Interactive Filters

The dashboard provides interactive filters for content type, release year, and rating. Users can select different values to explore the Netflix dataset.

```python
content_type_filter = widgets.Dropdown(
    options=['All'] + sorted(df_clean['type'].unique().tolist()),
    value='All',
    description='Content Type:'
)

release_year_filter = widgets.Dropdown(
    options=['All'] + sorted(df_clean['release_year'].unique().tolist(), reverse=True),
    value='All',
    description='Release Year:'
)

rating_filter = widgets.Dropdown(
    options=['All'] + sorted(df_clean['rating'].unique().tolist()),
    value='All',
    description='Rating:'
)

display(
    content_type_filter,
    release_year_filter,
    rating_filter
)
```

### 8.3 KPI Summary

Key Performance Indicators (KPIs) are displayed to provide a quick summary of the filtered Netflix content, including total titles, movies, TV shows, and average movie duration.

```python
def get_filtered_data():
    filtered = df_clean.copy()

    if content_type_filter.value != 'All':
        filtered = filtered[
            filtered['type'] == content_type_filter.value
        ]

    if release_year_filter.value != 'All':
        filtered = filtered[
            filtered['release_year'] == release_year_filter.value
        ]

    if rating_filter.value != 'All':
        filtered = filtered[
            filtered['rating'] == rating_filter.value
        ]

    return filtered


def show_kpis():
    filtered = get_filtered_data()

    total_titles = len(filtered)
    movies = (filtered['type'] == 'Movie').sum()
    tv_shows = (filtered['type'] == 'TV Show').sum()

    movie_durations = pd.to_numeric(
        filtered.loc[
            filtered['type'] == 'Movie',
            'duration'
        ].str.extract(r'(\d+)')[0],
        errors='coerce'
    )

    avg_duration = movie_durations.mean()

    print("========== NETFLIX DASHBOARD KPIs ==========")
    print("Total Titles:", total_titles)
    print("Movies:", movies)
    print("TV Shows:", tv_shows)
    print(
        "Average Movie Duration:",
        round(avg_duration, 2) if pd.notna(avg_duration) else "N/A",
        "minutes"
    )

show_kpis()
```

### 8.4 Interactive Charts

The dashboard displays interactive charts based on the selected filters. Users can hover over charts, zoom, and explore the filtered Netflix content.

```python
# Interactive Dashboard

def show_dashboard(content_type='All', release_year='All', rating='All'):

    filtered = df_clean.copy()

    # Content Type filter
    if content_type != 'All':
        filtered = filtered[filtered['type'] == content_type]

    # Release Year filter
    if release_year != 'All':
        filtered = filtered[filtered['release_year'] == int(release_year)]

    # Rating filter
    if rating != 'All':
        filtered = filtered[filtered['rating'] == rating]

    print("========== NETFLIX INTERACTIVE DASHBOARD ==========")
    print("Total Titles:", len(filtered))
    print("Movies:", (filtered['type'] == 'Movie').sum())
    print("TV Shows:", (filtered['type'] == 'TV Show').sum())

    # Stop if no records match
    if len(filtered) == 0:
        print("\nNo titles match the selected filters.")
        return

    # Chart 1
    type_counts = filtered['type'].value_counts()

    plt.figure(figsize=(7, 4))
    type_counts.plot(kind='bar')
    plt.title("Movies vs TV Shows")
    plt.xlabel("Content Type")
    plt.ylabel("Number of Titles")
    plt.xticks(rotation=0)
    plt.show()

    # Chart 2
    year_counts = filtered['release_year'].value_counts().sort_index()

    plt.figure(figsize=(12, 4))
    year_counts.plot()
    plt.title("Content by Release Year")
    plt.xlabel("Release Year")
    plt.ylabel("Number of Titles")
    plt.show()

    # Chart 3
    rating_counts = filtered['rating'].value_counts().head(10)

    plt.figure(figsize=(10, 5))
    rating_counts.sort_values().plot(kind='barh')
    plt.title("Top Ratings")
    plt.xlabel("Number of Titles")
    plt.ylabel("Rating")
    plt.show()


# Reset filters to All
content_type_filter.value = 'All'
release_year_filter.value = 'All'
rating_filter.value = 'All'

dashboard = widgets.interactive(
    show_dashboard,
    content_type=content_type_filter,
    release_year=release_year_filter,
    rating=rating_filter
)

display(dashboard)
```
